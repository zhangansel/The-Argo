```python
print(1,2,3,4)#输出1 2 3 4 默认分隔符为空格
print(1,2,3,4,sep="->")#sep把分隔符改为了箭头
print(1234)
print(1234,sep="--")
#由此可见，逗号只是一个切分点，我要是直接输入1234，会被识别为一个数字，就谈不上分隔了，自然不会显示分隔符
```

```python
print(1,2,3)
print(1,2,3)

print(1,2,3,end="\t")#\t为制表符，原本是默认换行的，用了之后不换行了，相当于三个空格（一个tab）的大小
print(1,2,3)
```

```python
#打开名为‘报错’的文件，以追加模式（“a”）打开，W则为写入，会覆盖原文本，也就是原文本全都没了换成新写的
#如果文件不存在会自己创建一个新的文件
f = open("报错.txt","a")
print("hello world ",file=f)
#将字符串“hello world”写入文件中，使用print函数，并指定输出目标为f文件
print("print语句需要加上扩号",file=f)
f.close()#关闭文件，确保所有写入的内容保存到文件中
```

# 格式化输出
## 给定一个模板，按照模板输出内容
## 格式化输出使用%和不同的字符连用，不同类型的数据需要使用不同的格式化字符

## 常用格式化字符及其含义
%c   字符         eg:‘c’ 等单字符
%s   字符串       eg:"abc" 等双引号定义的
%d   十进制整数   eg：1，2，3 等整数
%f   浮点数       eg:1.2 , 1.3 等等小数

```python
#基本格式化输出

print("我的名字是%s,年龄是%d岁"%("01",18))
print("我的名字是%s,年龄是%d岁"%("Ansel",18))
print("我的名字是%s,年龄是%d岁"%("蝙蝠侠",18))
print("我的名字是%s,年龄是%d岁"%("01","18"))#注意引号问题
```
